# EZ

```python
### This combines one group
## EZ

# Get the filenames of the uploaded PDFs
pdf_dir = '/content'  # Colab's default directory for file uploads
pdf_files = [filename for filename in os.listdir(pdf_dir) if filename.lower().endswith('.pdf')]

# Define the custom order of suffixes (last two characters of filenames)
custom_order = ['kg', 'D5', 'AR', '58', 'SV', 'DV']

# Normalize all to lowercase for comparison (optional)
custom_order_lower = [key.lower() for key in custom_order]

# Sort PDFs by the first 7 characters of the filename
pdf_files.sort(key=lambda x: x[:7].lower())  # Sorting by first 7 chars

# Group PDFs by the last two characters
suffix_map = {key.lower(): [] for key in custom_order}

for filename in pdf_files:
    suffix = filename[:-4][-2:].lower()  # Get the last 2 characters before ".pdf"
    if suffix in suffix_map:
        suffix_map[suffix].append(os.path.join(pdf_dir, filename))


# VALIDATION: ensure all required files exist
missing_suffixes = [key.upper() for key, files_list in suffix_map.items() if not files_list]

if missing_suffixes:
    raise FileNotFoundError(
        f"Missing required PDF files for suffix(es): {', '.join(missing_suffixes)}"
    )


# Merge PDFs in the given custom order
merger = PdfMerger()
for key in custom_order_lower:
    for pdf_path in suffix_map.get(key, []):
        merger.append(pdf_path)

# Determine the new filename based on the DV-suffixed file
dv_file = next((filename for filename in pdf_files if filename.endswith('DV.pdf')), None)

if dv_file:
    # Remove "DV" and add "2525"
    base_name = dv_file[:-6]  # Remove the last 2 characters "DV" and ".pdf"
    output_filename = f"{base_name}2525.pdf"
    output_sigs = f"{base_name}Sigs.pdf"
else:
    # Default name if no "DV" file found (just in case)
    output_filename = "2525.pdf"
    output_sigs = "Sigs.pdf"

# Save the merged PDF with the new filename
output_path = os.path.join(pdf_dir, output_filename)
with open(output_path, 'wb') as f_out:
    merger.write(f_out)
merger.close()


# Provide the merged PDF for download
files.download(output_path)

print(f"PDFs sorted, merged, and saved as {output_filename}. Downloading the merged file now.")


## Extract the signature pages
if len(output_filename) >= 10 and output_filename[-10:-8] == "DB":
    extract_pages = [1, 2]  # Pages 2 and 3
else:
    extract_pages = [1] # Page 2

# Read source PDF
reader = PdfReader(output_path)
writer = PdfWriter()

# Add selected pages if they exist
for page_num in extract_pages:
    if page_num < len(reader.pages):
        writer.add_page(reader.pages[page_num])

# Create output file
output_pdf = os.path.join(
    os.path.dirname(output_path),
    output_sigs
)

with open(output_pdf, "wb") as f:
    writer.write(f)

print(f"Created: {output_pdf}")

# Download the file
files.download(output_pdf)

## Deletes Files

# Get the filenames of the uploaded PDFs
pdf_dir = '/content'  # Colab's default directory for file uploads

# List all files in the directory
files_in_directory = os.listdir(pdf_dir)

# Loop through each file and remove it
for filename in files_in_directory:
    if filename == output_filename or filename == output_sigs:
      continue

    file_path = os.path.join(pdf_dir, filename)
    # Check if it is a file (not a folder)
    if os.path.isfile(file_path):
        os.remove(file_path)

print("All files in the directory have been deleted.")
```

# Replace SB Sig for EZ

```python
## Replace SB Signature page for EZ

# Find filenames
base_pdf = next((f for f in os.listdir() if f.endswith("2525.pdf")), None)
sig_pdf = next((f for f in os.listdir() if f.endswith("Sig.pdf")), None)

if not base_pdf or not sig_pdf:
    raise FileNotFoundError("Either the base or signature PDF file is missing.")

# Read PDFs
reader_base = PdfReader(base_pdf)
reader_sig = PdfReader(sig_pdf)
writer = PdfWriter()

# Page index: 2 = 3rd page (0-based index)
replace_index = 2

# Use first page of the signature PDF
sig_page = reader_sig.pages[0]

# Add pages, replacing page 3
for i in range(len(reader_base.pages)):
    if i == replace_index:
        writer.add_page(sig_page)
    else:
        writer.add_page(reader_base.pages[i])

# Overwrite the original file
with open(base_pdf, "wb") as f_out:
    writer.write(f_out)

print(f"Replaced page 3 and saved to: {base_pdf}")

files.download(base_pdf)
```
